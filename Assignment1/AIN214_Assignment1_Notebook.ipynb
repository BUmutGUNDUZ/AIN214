# Questions (50p)

1. Read the csv file and create database. (10p)
2. Find the top 3 restaurants with the highest total revenue. (revenue = quantity x price) (5p)
3. Calculate the total quantity of dishes sold by category for each restaurant. (5p)
4. List the top 5 customer with highest total spending (also show total spending value) (5p)
5. Find the average rating for each restaurant. (5p)
6. Check for duplicate orders (same customer_id+restaurant_id+order_date). (5p)

8. Analyze whether specific cities contribute a significantly larger portion to the total revenue compared to others. (8p)

```python
df = pd.read_csv("Data/Foodpanda Analysis Dataset.csv")

df["restaurant_id"] = df["restaurant_name"].factorize()[0] + 1

conn = sqlite3.connect(":memory:")

conn.executescript("""
DROP TABLE IF EXISTS Customer;
DROP TABLE IF EXISTS Restaurant;
DROP TABLE IF EXISTS Orders;

CREATE TABLE Customer (
    customer_id TEXT PRIMARY KEY,
    gender TEXT,
    age TEXT,
    city TEXT,
    signup_date TEXT,
    order_frequency INTEGER,
    last_order_date TEXT,
    loyalty_points INTEGER,
    churned TEXT
);

CREATE TABLE Restaurant (
    restaurant_id INTEGER PRIMARY KEY,
    restaurant_name TEXT
);

CREATE TABLE Orders (
    order_id TEXT PRIMARY KEY,
    customer_id TEXT,
    restaurant_id INTEGER,
    order_date TEXT,
    dish_name TEXT,
    category TEXT,
    quantity INTEGER,
    price REAL,
    payment_method TEXT,
    rating INTEGER,
    rating_date TEXT,
    delivery_status TEXT,
    FOREIGN KEY (customer_id) REFERENCES Customer(customer_id),
    FOREIGN KEY (restaurant_id) REFERENCES Restaurant(restaurant_id)
);
""")

df_customers = df[[
    "customer_id", "gender", "age", "city", "signup_date",
    "order_frequency", "last_order_date", "loyalty_points", "churned"
]].drop_duplicates(subset=["customer_id"])

df_restaurants = df[["restaurant_id", "restaurant_name"]].drop_duplicates(subset=["restaurant_id"])

df_orders = df[[
    "order_id", "customer_id", "restaurant_id", "order_date", "dish_name",
    "category", "quantity", "price", "payment_method",
    "rating", "rating_date", "delivery_status"
]]

df_customers.to_sql("Customer", conn, if_exists="append", index=False)
df_restaurants.to_sql("Restaurant", conn, if_exists="append", index=False)
df_orders.to_sql("Orders", conn, if_exists="append", index=False);
```

```python
#2
query = """
SELECT R.restaurant_name, SUM(O.quantity * O.price) AS total_revenue FROM Orders AS O
JOIN Restaurant AS R ON O.restaurant_id = R.restaurant_id
GROUP BY R.restaurant_id
ORDER BY total_revenue DESC
LIMIT 3;
"""

df_result = pd.read_sql_query(query, conn)
print(df_result)
```

```python
#3
query = """
SELECT R.restaurant_name, O.category, SUM(O.quantity) AS total_quantity FROM Orders AS O
JOIN Restaurant AS R ON O.restaurant_id = R.restaurant_id
GROUP BY R.restaurant_name, O.category
ORDER BY R.restaurant_name, total_quantity DESC;
"""

df_result = pd.read_sql_query(query, conn)
print(df_result)
```

```python
#4
query = """
SELECT C.customer_id, SUM(O.price) AS total_spending FROM Orders AS O
JOIN Customer AS C ON O.customer_id = C.customer_id
GROUP BY C.customer_id
ORDER BY total_spending DESC
LIMIT 5;
"""

df_result = pd.read_sql_query(query, conn)
print(df_result)
```

```python
#5
query = """
SELECT R.restaurant_name, AVG(O.rating) AS average_rating FROM Orders AS O
JOIN Restaurant AS R ON O.restaurant_id = R.restaurant_id
GROUP BY R.restaurant_id
ORDER BY average_rating DESC;
"""

df_result = pd.read_sql_query(query, conn)
print(df_result)
```

```python
#6
query = """
SELECT customer_id, restaurant_id, order_date, COUNT(*) AS tekrar_sayisi FROM Orders
GROUP BY customer_id, restaurant_id, order_date
HAVING COUNT(*) > 1;
"""

df_result = pd.read_sql_query(query, conn)
print(df_result)
```

```python
#7
query = """
SELECT delivery_status, COUNT(*) * 100.0 / (SELECT COUNT(*) FROM Orders) AS percentage FROM Orders
GROUP BY delivery_status
ORDER BY percentage DESC;
"""

df_result = pd.read_sql_query(query, conn)
print(df_result)
```

```python
#8
query = """
SELECT C.city, SUM(O.quantity * O.price) AS total_revenue, ROUND(SUM(O.quantity * O.price) * 100.0 /
(SELECT SUM(quantity * price) FROM Orders), 2) AS percentage_of_total
FROM Customer AS C
JOIN Orders AS O ON O.customer_id = C.customer_id
GROUP BY C.city
ORDER BY total_revenue DESC;
"""

df_result = pd.read_sql_query(query, conn)
print(df_result)
```
